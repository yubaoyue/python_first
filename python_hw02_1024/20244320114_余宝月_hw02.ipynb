字符串

```python
# 1. is_palindrome函数 判断回文
def is_palindrome(s: str) -> bool:
    # 忽略空格，全部转小写
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]

print("is_palindrome 测试")
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("hello world"))
```

```python
# 2. is_abecedarian 字母顺序（忽略空格）
def is_abecedarian(s: str) -> bool:
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True

print("is_abecedarian 测试")
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
```

```python
#3. remove_vowels 删除元音 a e i o u
def remove_vowels(s: str) -> str:
    vowels = {"a","e","i","o","u","A","E","I","O","U"}
    res = []
    for ch in s:
        if ch not in vowels:
            res.append(ch)
    return "".join(res)

print("remove_vowels 测试")
print(repr(remove_vowels('cat')))
print(repr(remove_vowels('audio')))
print(repr(remove_vowels("aeiou")))
```

若全部为元音字母，则返回空字符串

```python
#1 list_reverse，输入不是列表抛异常
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

print("list_reverse 测试")
print(list_reverse([1,2,3]))
print(list_reverse(["a","b","c"]))

#2 is_sorted，序列升序，只遍历一次
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

print("is_sorted 测试")
print(is_sorted([1,2,3,4]))
print(is_sorted([4,3,2,1]))
print(is_sorted((1,3,5,7))) #也可以传元组
```

元组

```python
#1 vec_scalar_mult 元组向量标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须是int/float")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是int/float")
    return tuple(x * s for x in t)

print("vec_scalar_mult 测试")
print(vec_scalar_mult((1,2,3),2))
print(vec_scalar_mult((2.0,4.0), 0.5))
```

```python
#2 vec_inner_product 向量内积
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入必须为元组")
    if len(x) != len(y):
        raise ValueError("两个向量元组长度不一致")
    total = 0.0
    for xi, yi in zip(x,y):
        if not (isinstance(xi,(int,float)) and isinstance(yi,(int,float))):
            raise TypeError("向量元素必须是数字")
        total += xi * yi
    return total

print("vec_inner_product 测试")
print(vec_inner_product((1,2),(3,4)))
```

```python
#3 check_valid_mx 检查有效矩阵：元组的元组
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    row_len = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_len is None:
            row_len = len(row)
        else:
            if len(row) != row_len:
                return False
        for val in row:
            if not isinstance(val, (int, float)):
                return False
    return True

print("check_valid_mx 测试")
good_mx = ((1,2),(3,4))
bad_mx1 = ((1,2), (3,4,5))
print(check_valid_mx(good_mx))
print(check_valid_mx(bad_mx1))
```

```python
#4 mx_vec_mult 矩阵乘向量
def mx_vec_mult(mat, vec):
    if not check_valid_mx(mat):
        raise TypeError("不是合法矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(mat[0])
    if len(vec) != n_col:
        raise ValueError("矩阵列数与向量维度不匹配")
    result = []
    for row in mat:
        dot = vec_inner_product(row, vec)
        result.append(dot)
    return tuple(result)

print("mx_vec_mult 测试")
M = ((1,2),(3,4))
v = (10,20)
print(mx_vec_mult(M, v))
```

面向对象的方法

```python
class Vector:
    def __init__(self, dim, elements=None):
        #dim:维度（非负整数）， elements:列表/元组，向量元素；可以不传
        # 检查维度
        if not isinstance(dim, int):
            raise TypeError("维度dim必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim  #实例属性
        if elements is None:
            #不传元素，默认全部0
            self.elements = [0.0 for _ in range(dim)]
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements必须list或者tuple")
            if len(elements) != dim:
                raise ValueError(f"维度{dim}和元素数量{len(elements)}不匹配")
            #全部转float保存
            self.elements = [float(x) for x in elements]

    def __eq__(self, other):
        """ == 相等判断 """
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements == other.elements

    def __lt__(self, other):
        """ < 小于，字典序比较 """
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements

    def __le__(self, other):
        """ <= """
        return self < other or self == other

    def __gt__(self, other):
        """ > """
        return not self <= other

    def __ge__(self, other):
        """ >= """
        return not self < other

    def dot(self, other):
        """向量内积方法"""
        if not isinstance(other, Vector):
            raise TypeError("dot的参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，无法计算内积")
        s = 0.0
        for a,b in zip(self.elements, other.elements):
            s += a * b
        return s

    def __mul__(self, other):
        """ v * x；x可以是标量或者Vector对象 """
        # 向量 * 标量
        if isinstance(other, (int, float)):
            new_elems = [e * other for e in self.elements]
            return Vector(self.dim, new_elems)
        # 向量 * 向量，逐元素相乘
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量维度必须相同才能相乘")
            new_elems = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("*操作只支持数字或者Vector对象")

    def __rmul__(self, scalar):
        """ 标量 * v，例如2*v """
        return self * scalar

    def __repr__(self):
        return f"Vector(dim={self.dim}, elements={self.elements})"

print("Vector类测试")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
print("v1 =", v1)
print("v2 =", v2)
print("v1 * 2 →", v1 * 2)
print("2 * v1 →", 2 * v1)
print("v1 * v2 →", v1 * v2)
print("v1.dot(v2) →", v1.dot(v2))
print("v1 < v2 →", v1 < v2)
print("v1 == Vector(2,(2,3)) →", v1 == Vector(2,(2,3)))

# 默认全零向量
v_zero = Vector(3)
print("默认零向量 Vector(3):", v_zero)
```

问题2：维度dim 和 elements 应该是实例属性：
    应该使用实例属性。
    每一个Vector实例代表一个独立向量，不同向量可以拥有不同维度、不同元素；
    如果作为类属性，则所有对象共享同一套维度和数据，创建多个向量时会互相覆盖，无法独立保存每个向量自己的数据。
