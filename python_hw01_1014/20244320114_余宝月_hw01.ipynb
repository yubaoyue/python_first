欧几里得算法

```python
def gcd(m, n):
    while n != 0:
        remainder = m % n   # m除以n得到余数
        m = n               # 把n赋值给m
        n = remainder       # 余数赋值给n
    # 当n等于0的时候，m就是最大公约数
    return m

# 测试三组数字
print("gcd(2018, 2019) =", gcd(2018,2019))
print("gcd(1200, 300) =", gcd(1200,300))
print("gcd(5040, 60) =", gcd(5040,60))

# 测试负数，看效果
print("\n测试负数输入")
print("gcd(-12, 8) =", gcd(-12,8))
```

Python的%取模，结果符号与除数保持一致。如果参数是一个或者两个负数，函数返回负数，但是最大公约数应当大于等于0，所以这样没有数学意义。

近似欧拉数e

```python
import math

# 1. euler_limit：(1+1/n)**n
def euler_limit(n):
    val = (1 + 1 / n) ** n
    return float(val)

# 2. euler_infinite_sum(n) 返回前n项的和
def euler_infinite_sum(n):
    total = 0.0
    for k in range(n):
        # k从0开始，每一项1/(k!)
        fact = math.factorial(k)
        total = total + 1.0 / fact
    return total

# 3. euler_approx(ϵ)，直到误差小于epsilon
def euler_approx(epsilon):
    real_e = math.exp(1)
    total = 0.0
    k = 0
    while True:
        term = 1.0 / math.factorial(k)
        total += term
        diff = abs(total - real_e)
        if diff < epsilon:
            break
        k += 1
    return total

#4. print_euler_sum_table(n) k从1到n打印每一个结果
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")


# 测试
print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("真实e:", math.exp(1))
print("euler_approx(0.001):", euler_approx(0.001))
print("\n表格测试")
print_euler_sum_table(5)
```

测试整数属性

```python
def is_power_of_2(n):
    # n是正整数，不断除以2，看最后是不是等于1
    while n > 1:
        if n % 2 != 0:
            # 不能被2整除，不是2的幂
            return False
        n = n // 2  # 整数除法
    return True

def is_power(b, n):
    # 判断n是不是b的幂，参数顺序 b, n
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

# 测试
print("is_power_of_2(8):", is_power_of_2(8))   # True
print("is_power_of_2(10):", is_power_of_2(10)) # False
print("is_power_of_2(1):", is_power_of_2(1))   # True

print("\nis_power测试：")
print("is_power(3, 9):", is_power(3,9)) # True
print("is_power(5,13):", is_power(5,13)) # False
```
