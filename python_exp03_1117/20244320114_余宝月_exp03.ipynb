实验1 正则表达式的基础匹配

```python
import re
```

任务1 匹配文本中数字
\d+ 匹配1个以上数字
\s+ 匹配1个以上空白字符(空格、换行)
\bcat\b 匹配完整单词cat，不会匹配cats里面的cat

```python
text1 = "cat on mat; later: cats and dogs"

res_d = re.findall(r"\d+", text1)

res_s = re.findall(r"\s+", text1)

res_cat_word = re.findall(r"\bcat\b", text1)

print("\\d+匹配结果：", res_d)
print("\\s+匹配结果：", res_s)
print(r"\bcat\b匹配结果：", res_cat_word)
```

任务2 差异

```python
text2 = "cat dog bird CAT"
res_or = re.findall(r"cat|dog", text2)
print("cat|dog匹配：", res_or)

# 分支顺序差异
r1 = re.findall(r"a|aa|aaa", "aaaa")
r2 = re.findall(r"aaa|aa|a", "aaaa")
print("a|aa|aaa 结果", r1)
print("aaa|aa|a 结果", r2)
```

正则中 |  从左到右，匹配成功就不再试后面的

任务3 

```python
text3 = "ct cat caat caaat"
print("原始文本：", text3)
print("ca*t ：", re.findall(r"ca*t", text3))
print("ca+t ：", re.findall(r"ca+t", text3))
print("ca?t ：", re.findall(r"ca?t", text3))
print("ca{2}t：", re.findall(r"ca{2}t", text3))
print("ca{1,2}t：", re.findall(r"ca{1,2}t", text3))
```

ca*t 表示a出现0次或者多次
ca+t 表示a至少出现1次
ca?t 表示a出现0次或者1次
ca{2}t 表示a出现2次
ca{1,2}t 表示a出现1次和2次

实验2 分组、回溯引用与替换

任务1 提取邮箱地址的用户名和域名

```python
email_text = "联系:alice@example.com;备选:bob@test.org"
# 分组 第一组 用户名在@前，第二组 域名在@后面
pattern_email = r"([a-zA-Z0-9]+)@([a-zA-Z0-9.]+)"
emails = re.findall(pattern_email, email_text)
print("(用户名,域名):", emails)
for user, domain in emails:
    print(f"用户名:{user}  域名:{domain}")
```

任务2 后向引用找连续重复单词

```python
dup_text = "cat cat, dog cat, bird bird"

dup_pattern = r"\b(\w+)\s+\1\b"
repeat_words = re.findall(dup_pattern, dup_text)
print("连续重复出现的单词：", repeat_words)
```

任务3 转换形式

```python
date_text = """2024-05-01
2024/05/02
2024.05.03"""
print("转换前")
print(date_text)
# 捕获三组：分隔符可以 - / .
date_pat = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"

date_result = re.sub(date_pat, r"\2-\3-\1", date_text)
print("\n转换后")
print(date_result)
```

任务4 电话号码格式转换

```python
phone_text = "联系我:123-456-7890 或 987.654.3210"
print("转换前：", phone_text)
phone_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
phone_out = re.sub(phone_pat, r"(\1)\2-\3", phone_text)
print("转换后：", phone_out)
```

实验3 使用BeautifulSoup解析HTML

```python
from bs4 import BeautifulSoup
```

```python
html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""
```

任务1 构建解析树并输出

```python
soup = BeautifulSoup(html_doc, "html.parser")

title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())
```

任务2 find_all()找class="course"，输出链接文字和href属性

```python
p_list = soup.find_all("p", class_="course")
print("\nclass=course段落")
for p in p_list:
    print(p.get_text())

a_list = soup.find_all("a")
print("\n所有链接")
for a in a_list:
    link_text = a.get_text()
    link_href = a["href"]
    print(f"链接文字：{link_text}，href={link_href}")
```

任务3 get_text()提取页面可见文本

```python
print("\n整个文档可见文本")
print(soup.get_text())

print("\n第一个course段落文本")
print(p_list[0].get_text())

print("\n第一个a链接文本")
print(a_list[0].get_text())

print("\n第二个course段落文本")
print(p_list[1].get_text())

print("\n第二个a链接文本")
print(a_list[1].get_text())
```
