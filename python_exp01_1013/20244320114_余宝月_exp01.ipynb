实验2：函数、条件、循环与递归

```python
#任务1
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

#测试样例
print(score_level(95))
print(score_level(82))
print(score_level(70))
print(score_level(50))

#测试越界，捕获异常
try:
    score_level(105)
except ValueError as err:
    print("越界测试报错：", err)

print("\n")
#任务2
#for版本
sum_for = 0
for i in range(1,101):
    sum_for += i
print("for循环1‑100和：", sum_for)

#while版本
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i +=1
print("while循环1‑100和：", sum_while)

print("\n")
#任务3
#迭代版阶乘
def fact_iter(n):
    if n <0:
        raise ValueError("只能输入非负整数")
    res =1
    for k in range(2, n+1):
        res *=k
    return res

#递归版阶乘
def fact_recur(n):
    if n <0:
        raise ValueError("只能输入非负整数")
    if n ==0 or n ==1:
        return 1
    return n * fact_recur(n-1)

#测试
for n in [0,1,5,10]:
    print(f"n={n},迭代阶乘={fact_iter(n)},递归阶乘={fact_recur(n)}")

print("\n")
#任务4
def summary(values):
    if len(values)==0:
        raise ValueError("列表不能为空")
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

#测试，元组解包
lst = [85,58,92,76,60,89]
mi,ma,av = summary(lst)
print(f"最小:{mi},最大:{ma},平均:{av:.2f}")

print("\n")
#任务5
def sum_all(*args):
    return sum(args)

print("无参数：", sum_all())
print("单个参数：", sum_all(10))
print("多个参数：", sum_all(1,2,3,4,5))
```

实验4：文件读写、异常处理与模块

```python
#任务1
import os

cwd = os.getcwd()
print("当前工作目录：", cwd)

data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录路径：", data_dir)
print("\n")
#任务2
file_path = os.path.join(data_dir, "students.txt")

content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

#写入文件
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content_text)

#读取文件
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("全部读取行：")
for line in lines:
    print(repr(line))
print("\n")
#任务3
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) !=2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if len(name)==0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩'{score_str}'不是合法整数")
    if not (0 <= score <=100):
        raise ValueError(f"分数{score}超出0‑100范围")
    return {"name":name, "score":score}
print("\n")
#任务4
valid_records = []
error_logs = []

#跳过第一行表头
for line_no, raw_line in enumerate(lines, start=1):
    if line_no ==1:
        continue
    try:
        res = parse_student(raw_line)
        valid_records.append(res)
    except ValueError as err:
        error_logs.append( (line_no, str(err)) )

print("✅有效记录：")
for r in valid_records:
    print(r)

print("\n❌错误行日志(行号，错误原因)：")
for ln,msg in error_logs:
    print(f"行{ln}: {msg}")
print("\n")
#任务5
report_path = os.path.join(data_dir, "report.txt")

valid_num = len(valid_records)
invalid_num = len(error_logs)
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_count = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate = pass_count / valid_num
else:
    avg_score = 0
    pass_rate = 0

#w模式：覆盖原有文件，全部重写
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_num}\n")
    f.write(f"无效记录数：{invalid_num}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

# a追加模式：在文件末尾新增文字，不会删掉旧内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("===这是a模式追加进去的一行文字===\n")

print("report.txt已生成；w会覆盖全部旧内容；a仅在末尾追加")
print("\n")
#任务6
import pickle
pkl_path = os.path.join(data_dir, "students.pkl")

#dump保存
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

#load读取
with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)

print("pickle读取回来的数据：", loaded_data)
print("和原始有效记录是否相等：", loaded_data == valid_records)
print("\n")
#任务7
from student_utils import parse_student

#有效输入
print(parse_student("Tom,78"))

#无效输入
try:
    parse_student("Tom,200")
except ValueError as e:
    print("无效输入报错：", e)
print("\n")
#任务8
no_file = os.path.join(data_dir, "not_exist.txt")
try:
    with open(no_file,"r",encoding="utf-8") as f:
        pass
except FileNotFoundError:
    print(f"捕获异常：文件 {no_file} 不存在")
```

实验5：类、继承与对象复制

```python
import copy
#任务1
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >=60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
print("\n")
#任务2
for rec in valid_records:
    s_obj = Student(rec["name"], rec["score"])
    print(s_obj, "  是否及格：", s_obj.is_passed())
print("\n")
#任务3
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},research_topic={self.research_topic})"

#测试
g_stu = GraduateStudent("LiHua",88,"大模型应用研究")
print(g_stu)
print("研究生是否及格", g_stu.is_passed())
print("\n")
#任务4
#原始对象
origin = {"name": "Alice", "scores": [85, 90]}

#1直接赋值
obj1 = origin
obj1["scores"][0] = 999
print("直接赋值：")
print("origin:", origin)
print("obj1:", obj1)
#恢复原始
origin = {"name": "Alice", "scores": [85, 90]}

#2 copy.copy 浅拷贝
obj2 = copy.copy(origin)
obj2["scores"][0] = 999
print("\ncopy.copy()浅拷贝：")
print("origin:", origin)
print("obj2:", obj2)
origin = {"name": "Alice", "scores": [85, 90]}

#3 copy.deepcopy深拷贝
obj3 = copy.deepcopy(origin)
obj3["scores"][0] = 999
print("\ncopy.deepcopy()深拷贝：")
print("origin:", origin)
print("obj3:", obj3)
```
